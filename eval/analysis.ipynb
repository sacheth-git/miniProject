# SkillMatch evaluation results

Loads the files written by `eval/evaluate.py`, `latency.py`, `bias_check.py` and `shortlist_quality.py`. Run those first (see `eval/README.md`).

```python
import json
from pathlib import Path
from IPython.display import Markdown, Image, display

RESULTS = Path('results')
summary = json.loads((RESULTS / 'summary.json').read_text(encoding='utf-8'))
print('Labels:', summary['labels'])
print('Resumes:', summary['resumes'], 'Jobs in pool:', summary['jobs'])
```

```python
for name in ['main_table.md', 'significance.md', 'ablation.md']:
    display(Markdown((RESULTS / name).read_text(encoding='utf-8')))
```

```python
display(Image(str(RESULTS / 'ndcg_by_method.png')))
display(Image(str(RESULTS / 'ablation.png')))
```

```python
import pandas as pd
main = pd.read_csv(RESULTS / 'main_table.csv')
main.pivot(index='method', columns='metric', values='mean').round(3)
```
